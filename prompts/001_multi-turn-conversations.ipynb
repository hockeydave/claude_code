```python
# Load env variables
from dotenv import load_dotenv
load_dotenv()

# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5"

```

```python
# Make a single request to the API using a helper function
def send_prompt(client, model, user_prompt, max_tokens=1000):
    """Send a single-user prompt to Anthropic client and return response text."""
    message = client.messages.create(
        model=model,
        max_tokens=max_tokens,
        messages=[
            {
                "role": "user",
                "content": user_prompt
            }
        ],
    )
    return message.content[0].text

# Multi-turn conversation functions
def add_user_message(messages, user_prompt):
    """Add a user message to the messages list."""
    messages.append({
        "role": "user",
        "content": user_prompt
    })

def add_assistant_message(messages, assistant_response):
    """Add an assistant message to the messages list."""
    messages.append({
        "role": "assistant",
        "content": assistant_response
    })

def chat(messages):
    """Send a multi-turn conversation to the Anthropic client and return the assistant's response."""
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages,
    )
    return message.content[0].text
```

```python
# Example usage
user_prompt = "Define quantum computing in one sentence."
response_text = send_prompt(client, model, user_prompt)
response_text
```

```python
user_prompt = "Write another sentence"
response_text = send_prompt(client, model, user_prompt)
response_text
```

```python
# Create initial messages list for multi-turn conversation
messages = []

# Add in th einitial user question about quantum computing
add_user_message(messages, "Define quantum computing in one sentence.")

# Pass the list of messages into 'chat' to get a response
assistant_response = chat(messages)
assistant_response

add_assistant_message(messages, assistant_response)

# Add in the next user question
add_user_message(messages, "Write another sentence")

# Pass the updated list of messages into 'chat' to get a response
assistant_response = chat(messages)
assistant_response
```

```python
# Make an initial list of messages
messages = []

# Use a while True loop to allow for continuous conversation
while True:
    # Get user input
    user_prompt = input("You: ")
    
    # Add the user message to the messages list
    add_user_message(messages, user_prompt)
    
    # Call Claude and get the assistant's response
    assistant_response = chat(messages)
    
    # Print the assistant's response
    print(f"Assistant: {assistant_response}")
    
    # Add the assistant's response to the messages list
    add_assistant_message(messages, assistant_response)
```
